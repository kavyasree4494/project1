```python
import sqlite3
import os

if os.path.exists("library.db"):
    os.remove("library.db")
conn=sqlite3.connect("library.db")
cursor=conn.cursor()
cursor.execute("PRAGMA foreign_keys=ON;")
cursor.execute("""
CREATE TABLE IF NOT EXISTS Books(
book_id INTEGER PRIMARY KEY AUTOINCREMENT,
book_title TEXT,
author_name TEXT
)
""")
cursor.execute("""
CREATE TABLE IF NOT EXISTS Members(
member_id INTEGER PRIMARY KEY AUTOINCREMENT,
member_name TEXT,
member_email TEXT
)
""")
cursor.execute("""
CREATE TABLE IF NOT EXISTS loans(
loan_id INTEGER PRIMARY KEY AUTOINCREMENT,
book_id INTEGER,
member_id INTEGER,
loan_date TEXT,
FOREIGN KEY (book_id) REFERENCES Books(book_id),
FOREIGN KEY (member_id) REFERENCES Members(member_id)
)
""")
cursor.execute("INSERT INTO Books VALUES(101,'Python','John')")
cursor.execute("INSERT INTO Books VALUES(102,'Java','James')")
cursor.execute("INSERT INTO Books VALUES(103,'DA','David')")
cursor.execute("INSERT INTO Books VALUES(104,'ML','Michael')")

cursor.execute("INSERT INTO Members VALUES(201,'Arun','harrison@gmail.com')")
cursor.execute("INSERT INTO Members VALUES(202,'Ajay','ajay@gmail.com')")
cursor.execute("INSERT INTO Members VALUES(203,'Bala','bala@gmail.com')")
cursor.execute("INSERT INTO Members VALUES(204,'Daniel','daniel@gmail.com')")

cursor.execute("INSERT INTO loans VALUES(301,101,201,'2023-01-01')")
cursor.execute("INSERT INTO loans VALUES(302,102,202,'2023-02-01')")
cursor.execute("INSERT INTO loans VALUES(303,103,203,'2023-03-01')")
cursor.execute("INSERT INTO loans VALUES(304,104,204,'2023-04-01')")

cursor.execute("SELECT * FROM Books")
for row in cursor.fetchall():
    print(row)
cursor.execute("SELECT * FROM Members")
for row in cursor.fetchall():
    print(row)
cursor.execute("SELECT * FROM loans")
for row in cursor.fetchall():
    print(row)

cursor.execute("""
SELECT
Books.book_title,
Members.member_name
FROM loans
JOIN Books
ON loans.book_id=Books.book_id
JOIN Members
ON loans.member_id=Members.member_id
""")
for row in cursor.fetchall():
    print(row)
conn.commit()
conn.close()
```

```python
import sqlite3
import os

if os.path.exists("Product.db"):
    os.remove("Product.db")

conn = sqlite3.connect("Product.db") # Corrected connect method
cursor = conn.cursor() # Corrected cursor object creation
cursor.execute("PRAGMA foreign_keys=ON;")

cursor.execute("""
CREATE TABLE IF NOT EXISTS customers(
id INTEGER PRIMARY KEY,
customer_name TEXT,
customer_city TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS orders(
id INTEGER PRIMARY KEY,
customer_id INTEGER,
order_date TEXT, -- Changed to TEXT for date format
FOREIGN KEY (customer_id) REFERENCES customers(id)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS orderitems(
id INTEGER PRIMARY KEY,
order_id INTEGER, -- Removed PRIMARY KEY from order_id, it's a foreign key
product_name TEXT,
quantity INTEGER,
price INTEGER,
FOREIGN KEY (order_id) REFERENCES orders(id)
-- Removed FOREIGN KEY (customer_id) as per instruction
)
""")

cursor.execute("INSERT INTO customers VALUES(101,'Arun','Chennai')")
cursor.execute("INSERT INTO customers VALUES(102,'Ajay','Bangalore')")
cursor.execute("INSERT INTO customers VALUES(103,'Bala','Chennai')")
cursor.execute("INSERT INTO customers VALUES(104,'Daniel','Bangalore')")

cursor.execute("INSERT INTO orders VALUES(201,101,'2026-09-19')") # Corrected customer_id and added closing quote
cursor.execute("INSERT INTO orders VALUES(202,102,'2026-09-20')") # Corrected customer_id and added closing quote
cursor.execute("INSERT INTO orders VALUES(203,103,'2026-09-21')") # Corrected customer_id and added closing quote
cursor.execute("INSERT INTO orders VALUES(204,104,'2026-09-22')") # Corrected customer_id and added closing quote

cursor.execute("INSERT INTO orderitems VALUES(301,201,'BOOK',2,3050)")
cursor.execute("INSERT INTO orderitems VALUES(302,202,'LAPTOP',1,50000)")
cursor.execute("INSERT INTO orderitems VALUES(303,203,'MOBILE',1,40000)") # Added missing quantity
cursor.execute("INSERT INTO orderitems VALUES(304,204,'TABLET',3,20000)")

cursor.execute("SELECT * FROM customers")
for row in cursor.fetchall():
    print(row)
cursor.execute("SELECT * FROM orders")
for row in cursor.fetchall():
    print(row)
cursor.execute("SELECT * FROM orderitems")
for row in cursor.fetchall():
    print(row)

cursor.execute("""
SELECT
  customers.customer_name,
  SUM(orderitems.quantity * orderitems.price) AS total_sales -- Added alias for clarity
FROM customers
JOIN orders  ON customers.id = orders.customer_id
JOIN orderitems ON orders.id = orderitems.order_id -- Corrected join condition
GROUP BY customers.customer_name
""")
for row in cursor.fetchall():
    print(row)
conn.commit()
conn.close()
```
